## 1. Dữ liệu và tiền xử lý

Dữ liệu có 9.130 mẫu của 5 thành phố trong giai đoạn 2021–2025. Mỗi mẫu tương ứng một thành phố trong một ngày.

Đầu vào gồm temperature_2m_max, temperature_2m_min, temperature_2m_mean, sunshine_duration và wind_speed_10m_max. Biến mục tiêu là precipitation_sum liên tục, không chuyển thành nhãn có mưa/không mưa.

Dùng $z=\ln(1+y)$ để giảm độ lệch của mục tiêu và xử lý được giá trị 0. Không đưa precipitation_sum vào X, bỏ precipitation_hours và weather_code vì chúng trực tiếp mô tả hiện tượng cần ước lượng. Thành phố không được mã hóa, ngày chỉ dùng chia tập, precipitation_sum là tổng giáng thủy trong ngày, đơn vị mm.

## 2. Mô hình Linear Regression có hệ số chặn

Mô hình có dạng $\hat z=b+XW$, với $z=\ln(1+y)$ và b là hệ số chặn. Thêm cột toàn số 1 vào X để tạo $\widetilde X=[\mathbf{1},X]$, gộp b và W vào vector $\theta$:

$$\hat z=\widetilde X\theta, \qquad E(\theta)=\frac{1}{2}\|z-\widetilde X\theta\|_2^2.$$

Khi ma trận $\widetilde X$ đủ hạng cột, nghiệm OLS là $\hat\theta=(\widetilde X^T\widetilde X)^{-1}\widetilde X^Tz$. Code dùng np.linalg.lstsq để giải bài toán bình phương tối thiểu, tránh nghịch đảo trực tiếp.

Hệ số chặn được lưu tại self._W[0]. Khi đánh giá, đổi ngược cả mục tiêu và dự đoán từ thang log1p về đơn vị gốc bằng expm1.

## 3. Huấn luyện và đánh giá

Chia theo thời gian cho toàn bộ thành phố: train 2021–2024 có 7.305 mẫu, test 2025 có 1.825 mẫu. Cách chia này kiểm thử trên các ngày mới của những thành phố đã xuất hiện trong train, không trộn ngẫu nhiên theo dòng.

Đánh giá trên thang giáng thủy gốc bằng:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(y_i-\hat y_i)^2}.$$
