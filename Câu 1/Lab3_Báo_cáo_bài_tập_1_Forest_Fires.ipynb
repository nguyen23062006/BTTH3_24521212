## 2. Mô hình Linear Regression có hệ số chặn

Mô hình có dạng $\hat z=b+XW$, với $z=\ln(1+area)$ và b là hệ số chặn. Thêm cột toàn số 1 vào X để tạo $\widetilde X=[\mathbf{1},X]$, gộp b và W vào vector $\theta$:

$$\hat z=\widetilde X\theta, \qquad E(\theta)=\frac{1}{2}\|z-\widetilde X\theta\|_2^2.$$

Khi ma trận $\widetilde X$ đủ hạng cột, nghiệm OLS là $\hat\theta=(\widetilde X^T\widetilde X)^{-1}\widetilde X^Tz$. Code dùng np.linalg.lstsq để giải bài toán bình phương tối thiểu, tránh nghịch đảo trực tiếp.

Hệ số chặn được lưu tại self._W[0]. Khi đánh giá, đổi ngược cả mục tiêu và dự đoán từ thang log1p về đơn vị gốc bằng expm1.
