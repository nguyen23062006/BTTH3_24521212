# Báo cáo bài tập 4: So sánh Linear Regression NumPy và Scikit-learn trên Forest Fires
## 1. Dữ liệu
Forest Fires có 517 mẫu và 12 đặc trưng đầu vào. Mục tiêu area là diện tích rừng bị cháy, đơn vị ha. Đổi month và day sang số, biến đổi mục tiêu bằng log1p như bài 1.

Trộn dữ liệu với seed 42, dùng cùng 414 mẫu train và 103 mẫu test cho cả hai mô hình.

## 2. Hai mô hình so sánh

Cả hai mô hình có hệ số chặn và học trên cùng dữ liệu, với $z=\ln(1+area)$:

$$\hat z=b+XW.$$

Mô hình tự xây dựng dùng np.linalg.lstsq để tìm nghiệm bình phương tối thiểu. Hệ số chặn được học bằng cách thêm cột toàn số 1 vào đầu vào.

Mô hình thư viện dùng LinearRegression của Scikit-learn, đặt fit_intercept=True để học hệ số chặn. Hai mô hình đều giải bài toán bình phương tối thiểu trên cùng 12 đặc trưng.

## 3. Huấn luyện và đánh giá

Huấn luyện từng mô hình trên cùng tập train và dự đoán trên cùng tập test. Dùng chung cách đánh giá: đổi ngược mục tiêu và dự đoán từ thang log1p về diện tích gốc bằng expm1, rồi tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(area_i-\widehat{area}_i)^2}.$$
RMSE có đơn vị ha.
