# Báo cáo bài tập 4: So sánh Linear Regression NumPy và Scikit-learn trên Weather
## 1. Dữ liệu

Dùng weather_daily.csv đã thu thập, gồm 9.130 mẫu của 5 thành phố trong giai đoạn 2021–2025. Năm đặc trưng đầu vào là temperature_2m_max, temperature_2m_min, temperature_2m_mean, sunshine_duration và wind_speed_10m_max.

Mục tiêu là precipitation_sum, tổng giáng thủy trong ngày, đơn vị mm. Biến đổi mục tiêu bằng log1p như các bài trước.

Cả hai mô hình sử dụng cùng tập train 2021–2024 gồm 7.305 mẫu và tập test 2025 gồm 1.825 mẫu.

## 2. Hai mô hình so sánh

Cả hai mô hình có hệ số chặn và học trên cùng dữ liệu, với $z=\ln(1+y)$:

$$\hat z=b+XW.$$

Mô hình tự xây dựng dùng np.linalg.lstsq để giải bài toán bình phương tối thiểu, học hệ số chặn bằng cách thêm cột toàn số 1 vào đầu vào.

Mô hình thư viện dùng LinearRegression của Scikit-learn, đặt fit_intercept=True để học hệ số chặn. Hai mô hình đều dùng cùng 5 đặc trưng và cùng biến mục tiêu.

## 3. Huấn luyện và đánh giá

Huấn luyện từng mô hình trên cùng tập train và dự đoán trên cùng tập test. Dùng chung cách đánh giá: đổi ngược mục tiêu và dự đoán từ thang log1p về đơn vị gốc bằng expm1, rồi tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(y_i-\hat y_i)^2}.$$

RMSE có đơn vị mm.
