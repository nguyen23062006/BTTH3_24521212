# Báo cáo bài tập 2: Chuẩn hóa dữ liệu Weather
## 1. Dữ liệu

Dùng weather_daily.csv đã thu thập, gồm 9.130 mẫu của 5 thành phố trong giai đoạn 2021–2025. Mỗi mẫu là dữ liệu của một thành phố trong một ngày.

Năm đặc trưng đầu vào là temperature_2m_max, temperature_2m_min, temperature_2m_mean, sunshine_duration và wind_speed_10m_max. Mục tiêu là precipitation_sum, tổng giáng thủy trong ngày, đơn vị mm. Biến đổi mục tiêu bằng log1p.

Cả hai trường hợp dùng cùng tập train 2021–2024 gồm 7.305 mẫu và tập test 2025 gồm 1.825 mẫu.

## 2. Chuẩn hóa

Chuẩn hóa toàn bộ 5 cột đặc trưng theo công thức:

$$x'_{ij}=\frac{x_{ij}-\mu_j}{\sigma_j}.$$

Trung bình $\mu_j$ và độ lệch chuẩn $\sigma_j$ chỉ được tính trên train rồi dùng lại cho test. NumPy dùng ddof bằng 0. Sau chuẩn hóa, các cột trên train có trung bình gần 0 và phương sai gần 1, test không bắt buộc thỏa điều kiện này.

Không chuẩn hóa mục tiêu hoặc cột hệ số chặn. Tập train đang dùng không có cột đặc trưng hằng.

## 3. Huấn luyện và đánh giá

Dùng chung mô hình Linear Regression bằng NumPy có hệ số chặn, giải bằng np.linalg.lstsq cho dữ liệu gốc và dữ liệu đã chuẩn hóa. Giữ nguyên các mẫu, đặc trưng và mục tiêu giữa hai lần chạy.

Đổi ngược mục tiêu và dự đoán từ thang log1p về đơn vị gốc bằng expm1, sau đó tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(y_i-\hat y_i)^2}.$$

RMSE có đơn vị mm.
