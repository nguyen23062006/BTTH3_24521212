# Báo cáo bài tập 2: Chuẩn hóa dữ liệu Forest
## 1. Dữ liệu

Forest Fires có 517 mẫu và 12 đặc trưng đầu vào. Biến mục tiêu area là diện tích rừng bị cháy, đơn vị ha. Đổi month và day sang số, biến đổi mục tiêu bằng log1p.

Trộn dữ liệu với seed 42, dùng cùng 414 mẫu train và 103 mẫu test cho cả 2 lần huấn luyện và đánh giá. Dữ liệu gốc trong phép so sánh là dữ liệu đã mã hóa tháng/ngày nhưng chưa chuẩn hóa đặc trưng.

## 2. Chuẩn hóa

Chuẩn hóa toàn bộ 12 cột đặc trưng theo công thức:

$$x'_{ij}=\frac{x_{ij}-\mu_j}{\sigma_j}.$$

Trung bình $\mu_j$ và độ lệch chuẩn $\sigma_j$ chỉ được tính trên train rồi dùng lại cho test. NumPy dùng ddof bằng 0 khi tính độ lệch chuẩn. Các cột trên train sau chuẩn hóa có trung bình gần 0 và phương sai gần 1, test không bắt buộc thỏa điều kiện này.

Không chuẩn hóa mục tiêu hoặc cột hệ số chặn. Tập train đang dùng không có cột đặc trưng hằng.

## 3. Huấn luyện và đánh giá

Dùng cùng mô hình Linear Regression bằng NumPy có hệ số chặn, giải bằng np.linalg.lstsq cho hai trường hợp: đầu vào gốc và đầu vào đã chuẩn hóa.

Đổi ngược mục tiêu và dự đoán từ thang log1p về diện tích gốc bằng expm1, sau đó tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(area_i-\widehat{area}_i)^2}.$$

RMSE có đơn vị ha.
