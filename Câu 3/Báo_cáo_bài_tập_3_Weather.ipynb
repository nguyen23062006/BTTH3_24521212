## 1. Dữ liệu

Dùng weather_daily.csv đã thu thập, gồm 9.130 mẫu của 5 thành phố trong giai đoạn 2021–2025. Đầu vào ban đầu gồm temperature_2m_max, temperature_2m_min, temperature_2m_mean, sunshine_duration và wind_speed_10m_max.

Mục tiêu là precipitation_sum, tổng giáng thủy trong ngày, đơn vị mm. Biến đổi mục tiêu bằng log1p.

Cả hai trường hợp sử dụng cùng tập train 2021–2024 gồm 7.305 mẫu và tập test 2025 gồm 1.825 mẫu.

## 2. Chọn đặc trưng

Tính ma trận tương quan Pearson chỉ trên X_train. Duyệt các cột theo thứ tự trong dữ liệu, giữ một cột nếu tương quan tuyệt đối của nó với mọi cột đã giữ nhỏ hơn 0,8. Nếu có $|r|\geq 0{,}8$, loại cột đang xét. Đây là cách giảm đa cộng tuyến gần đúng dựa trên tương quan từng cặp.

Trên train, temperature_2m_max có tương quan khoảng 0,846871 với temperature_2m_min và 0,949176 với temperature_2m_mean. Vì temperature_2m_max xuất hiện trước, giữ cột này và loại hai cột nhiệt độ còn lại. Số đặc trưng giảm từ 5 xuống 3, gồm temperature_2m_max, sunshine_duration và wind_speed_10m_max. Áp dụng cùng danh sách cột cho train và test.

## 3. Huấn luyện và đánh giá

Dùng chung mô hình Linear Regression bằng NumPy có hệ số chặn, giải bằng np.linalg.lstsq. So sánh mô hình dùng đủ 5 đặc trưng và mô hình dùng 3 đặc trưng sau xử lý. Các lựa chọn khác được giữ nguyên.

Đổi ngược mục tiêu và dự đoán từ thang log1p về đơn vị gốc bằng expm1, sau đó tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(y_i-\hat y_i)^2}.$$

RMSE có đơn vị mm.
