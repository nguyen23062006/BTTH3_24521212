# Báo cáo bài tập 3: Xử lý đa cộng tuyến trên Forest Fires
## 1. Dữ liệu
Forest Fires có 517 mẫu và 12 đặc trưng đầu vào. Mục tiêu area là diện tích rừng bị cháy, đơn vị ha. Đổi month và day sang số, biến đổi mục tiêu bằng log1p như các bài trước.

Trộn dữ liệu với seed 42, dùng cùng 414 mẫu train và 103 mẫu test để đánh giá cả hai trường hợp.

## 2. Chọn đặc trưng

Tính ma trận tương quan Pearson chỉ trên X_train. Duyệt các cột theo thứ tự trong dữ liệu, giữ một cột nếu tương quan tuyệt đối của nó với mọi cột đã giữ nhỏ hơn 0,8. Nếu có $|r|\geq 0{,}8$, loại cột đang xét.

Trên train, month và DC có tương quan khoảng 0,874735. Vì month xuất hiện trước, giữ month và loại DC. Số đặc trưng giảm từ 12 xuống 11, sau đó áp dụng cùng danh sách cột cho train và test.


## 3. Huấn luyện và đánh giá

Dùng chung mô hình Linear Regression bằng NumPy có hệ số chặn, giải bằng np.linalg.lstsq. So sánh mô hình dùng đủ 12 đặc trưng và mô hình dùng 11 đặc trưng sau khi loại DC. Các lựa chọn khác được giữ nguyên.

Đổi ngược mục tiêu và dự đoán từ thang log1p về diện tích gốc bằng expm1, rồi tính:

$$RMSE=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(area_i-\widehat{area}_i)^2}.$$

RMSE có đơn vị ha.
